### 1) Implementação do filtro FIR em hardware

O filtro FIR projetado no computador foi implementado no kit Cortex-M4 em duas representações numéricas:

- **Ponto flutuante (`float32`)**
- **Ponto fixo (`Q15`)**

O filtro foi projetado para atenuar a frequência de aproximadamente **2465 Hz**, utilizando uma frequência de amostragem de **48 kHz**, mesma frequência utilizada pelo kit de desenvolvimento.

Foram realizados testes utilizando ruído branco e um sinal de voz corrompido por uma componente aproximadamente constante em 2465 Hz. Os resultados foram analisados tanto no domínio da frequência, através da FFT, quanto no domínio tempo-frequência, através de espectrogramas.

#### Comparação entre ponto flutuante e ponto fixo

Não foi observada diferença expressiva entre as implementações em ponto flutuante e ponto fixo. Em ambos os casos, a atenuação da componente de aproximadamente 2465 Hz apresentou comportamento semelhante.

No teste com ruído branco, a componente próxima de 2465 Hz apresentava aproximadamente **-45 dB** antes da filtragem. Após a aplicação do filtro, essa componente foi reduzida para aproximadamente **-70 dB**, correspondendo a uma atenuação de aproximadamente **25 dB**.

Além da frequência central, o filtro apresentou uma região de atenuação ao redor de 2465 Hz. A atenuação começou a ser perceptível em aproximadamente **1500 Hz** e deixou de ser significativa próximo de **3900 Hz**.

#### Comparação com a implementação teórica em Python

Antes da implementação no Cortex-M4, o comportamento do filtro foi analisado no computador utilizando Python. Essa análise serve como referência teórica para comparação com os resultados obtidos no hardware.

Para o teste com ruído branco, a FFT do sinal original apresentou um nível espectral aproximadamente em **-60 dB** ao longo da faixa analisada. Na região da frequência de interesse, aproximadamente **2465 Hz**, a resposta do filtro apresentou uma forte atenuação, fazendo com que o nível nessa frequência chegasse a aproximadamente **-115 dB**.

Além da frequência central, o filtro apresentou uma região de rejeição ao seu redor. A atenuação começou a ser observada em aproximadamente **1300 Hz** e deixou de ser significativa próximo de **3700 Hz**.

Dessa forma, a implementação teórica indica que o filtro não atua somente exatamente em 2465 Hz, mas possui uma banda de rejeição ao redor dessa frequência. A maior atenuação ocorre na frequência central projetada.

Os resultados obtidos no hardware apresentaram o mesmo comportamento geral. No ruído branco, o nível espectral na região de 2465 Hz era aproximadamente **-45 dB** antes da filtragem e caiu para aproximadamente **-70 dB** após a aplicação do filtro. A região de atuação observada no hardware ficou aproximadamente entre **1500 Hz e 3900 Hz**.

Assim, embora os valores absolutos de amplitude e os limites da banda de rejeição apresentem algumas diferenças entre a implementação teórica e a implementação no Cortex-M4, as duas apresentam o mesmo comportamento qualitativo: uma forte atenuação centrada em aproximadamente **2465 Hz**, acompanhada por uma região de atenuação nas frequências próximas

| Característica | Python (teórico) | Cortex-M4 - Float | Cortex-M4 - Q15 |
|---|---:|---:|---:|
| Frequência central | ≈ 2465 Hz | ≈ 2465 Hz | ≈ 2465 Hz |
| Nível em 2465 Hz antes da filtragem | ≈ -60 dB | ≈ -45 dB | ≈ -45 dB |
| Nível em 2465 Hz após a filtragem | ≈ -115 dB | ≈ -70 dB | ≈ -70 dB |
| Início da atenuação | ≈ 1300 Hz | ≈ 1500 Hz | ≈ 1500 Hz |

## 2) Implementação do filtro IIR em hardware

O filtro IIR projetado no computador foi implementado no kit Cortex-M4 em **ponto flutuante**, utilizando uma cascata de estágios de segunda ordem (biquads).

O filtro foi projetado para atenuar a frequência de aproximadamente **2465 Hz**, utilizando uma frequência de amostragem de **48 kHz**, mesma frequência utilizada no kit de desenvolvimento.

Foram realizados testes utilizando ruído branco e um sinal de voz corrompido por uma componente aproximadamente constante em 2465 Hz. Os resultados foram analisados no domínio da frequência, através da FFT, e no domínio tempo-frequência, através de espectrogramas.

### Resultados com ruído branco

No teste realizado no hardware, o ruído branco apresentou um nível de aproximadamente **-45 dB** antes da filtragem. Após a aplicação do filtro IIR, a região próxima à frequência de interesse, em aproximadamente **2465 Hz**, foi atenuada para cerca de **-70 dB**.

A região de atenuação observada no hardware começa em aproximadamente **1600 Hz** e termina próxima de **3900 Hz**.

## 3) Implementação do filtro notch

O filtro notch foi implementado no kit Cortex-M4 como um filtro **IIR em ponto flutuante**, utilizando uma cascata de estágios de segunda ordem (biquad).

O objetivo do filtro é atenuar especificamente uma componente em aproximadamente **2465 Hz**, mantendo as demais frequências com pouca alteração. Dessa forma, o filtro notch apresenta uma região de rejeição muito mais estreita do que os filtros FIR e IIR analisados anteriormente.

A frequência de amostragem utilizada foi de **48 kHz**, tanto no kit de desenvolvimento quanto na implementação realizada em Python.

### Resultados com ruído branco

No teste utilizando ruído branco, a resposta do filtro não é tão facilmente visualizada na FFT devido à resolução utilizada no gráfico. A atenuação observada inicialmente parece relativamente pequena: a região próxima à frequência de interesse apresenta aproximadamente **-42 dB** antes da filtragem e cerca de **-52 dB** após a filtragem.

Apesar disso, é possível observar que a região afetada pelo filtro é bastante estreita. No hardware, a faixa de atenuação observada está aproximadamente entre **2200 Hz e 2800 Hz**.

O comportamento do filtro fica mais evidente quando aplicado ao áudio de voz contendo a componente de aproximadamente **2465 Hz**.

Essa forte atenuação também foi perceptível durante a reprodução do áudio. Diferentemente dos testes anteriores, a componente indesejada deixou de ser perceptível auditivamente após a aplicação do filtro.

### Comparação com a implementação em Python

Na implementação realizada em Python, o comportamento do filtro notch também apresentou uma rejeição bastante seletiva em torno da frequência de interesse.

No áudio corrompido, a componente de aproximadamente 2465 Hz apresentava inicialmente cerca de **-10 dB** e foi reduzida para aproximadamente **-90 dB** após a filtragem.

### Conclusão

O filtro notch apresentou uma característica diferente dos filtros FIR e IIR analisados anteriormente. Enquanto aqueles apresentavam uma região de atenuação mais ampla, o notch concentrou sua rejeição em uma faixa muito mais estreita ao redor da frequência de aproximadamente **2465 Hz**.

No teste com o áudio corrompido, o hardware reduziu a componente de interesse de aproximadamente **-28 dB para -95 dB**, tornando-a não perceptível auditivamente.

A implementação em Python apresentou comportamento semelhante, reduzindo a mesma componente de aproximadamente **-10 dB para -90 dB**. Assim, apesar das diferenças nos níveis absolutos e na largura aproximada da região de rejeição, os resultados confirmam o comportamento seletivo esperado do filtro notch.

# 4) Análise geral do projeto

Neste projeto foram implementados três filtros digitais no kit Cortex-M4 utilizando a biblioteca CMSIS-DSP: um filtro **FIR**, um filtro **IIR** em cascata de seções de segunda ordem (SOS) e um filtro **notch IIR**. Os filtros foram inicialmente projetados e analisados em Python e, posteriormente, implementados no hardware utilizando uma frequência de amostragem de **48 kHz**.

O objetivo principal foi atenuar uma interferência localizada em aproximadamente **2465 Hz**, permitindo comparar diferentes estruturas de filtros digitais e diferentes representações numéricas.

## Especificações dos filtros

### Filtro FIR

O filtro FIR implementado possui **152 coeficientes**. O projeto original possuía 151 coeficientes e foi acrescentado um coeficiente igual a zero ao final para atender à restrição da quantidade de coeficientes da implementação Q15 utilizada pelo CMSIS-DSP. Esse coeficiente adicional não altera a resposta do filtro.

Foram implementadas duas versões:

- **FIR em ponto flutuante (`float32`)**;
- **FIR em ponto fixo (`Q15`)**.

Os coeficientes das duas representações estão presentes no arquivo `coeffs_FIR.h`. A implementação em ponto flutuante utiliza os coeficientes como valores `float32`, enquanto a implementação Q15 utiliza os mesmos coeficientes quantizados para uma representação de 16 bits.

A resposta observada no hardware apresentou uma região de atenuação aproximadamente entre **1500 Hz e 3900 Hz**, com maior atenuação na região de interesse em torno de 2465 Hz.

### Filtro IIR

O filtro IIR possui **ordem 6** e foi implementado como uma cascata de **3 estágios de segunda ordem (SOS)**. Cada estágio possui cinco coeficientes utilizados pela implementação CMSIS-DSP, totalizando **15 coeficientes `float32`**.

Os coeficientes utilizados são:

```text
Estágio 1:
[b0, b1, b2] = [ 0.7442321, -1.4121811,  0.7442756]
[a1, a2]     = [ 1.6524835, -0.7418081]

Estágio 2:
[b0, b1, b2] = [ 1.0000000, -1.8974137,  1.0000006]
[a1, a2]     = [ 1.6160392, -0.8191712]

Estágio 3:
[b0, b1, b2] = [ 1.0000000, -1.8973892,  0.9999410]
[a1, a2]     = [ 1.8672226, -0.9114775]
```

No hardware, foi observada uma região de atenuação aproximadamente entre **1600 Hz e 3900 Hz**.

### Filtro notch IIR

O filtro notch é um filtro IIR de segunda ordem implementado utilizando apenas **uma seção de segunda ordem**, portanto possui somente cinco coeficientes no formato utilizado pelo CMSIS-DSP.

Os coeficientes utilizados são:

```text
[b0, b1, b2] = [ 1.0000000, -1.89678572, 1.0000000]
[a1, a2]     = [ 1.83355953, -0.93500000]
```

O filtro foi projetado especificamente para atenuar a frequência de aproximadamente **2465 Hz**. No hardware, a região de atenuação observada ficou aproximadamente entre **2200 Hz e 2800 Hz**, sendo mais estreita que a dos filtros FIR e IIR anteriores.

## Comparação dos resultados

| Filtro | Implementação | Estrutura | Região de atenuação observada no hardware |
|---|---|---|---|
| FIR | `float32` / Q15 | 152 coeficientes | ~1500–3900 Hz |
| IIR | `float32` | 3 SOS | ~1600–3900 Hz |
| Notch IIR | `float32` | 1 SOS | ~2200–2800 Hz |

Nos testes com o áudio corrompido por uma interferência em aproximadamente 2465 Hz, os três filtros foram capazes de reduzir significativamente a componente indesejada.

Para o FIR, a componente próxima de 2465 Hz passou aproximadamente de **-35 dB para -80 dB**. No IIR, foi observado comportamento semelhante. No filtro notch, a componente passou aproximadamente de **-28 dB para -95 dB**.

Também foi possível observar nos espectrogramas a redução da interferência após a filtragem. No caso do áudio filtrado pelo notch, a interferência deixou de ser perceptível auditivamente.

Os resultados obtidos no hardware apresentam o mesmo comportamento geral observado nas simulações em Python. Entretanto, os níveis absolutos dos espectros não são exatamente iguais entre as duas implementações. Por exemplo, nos testes com ruído branco, o nível de fundo observado no hardware foi diferente daquele obtido na simulação. Dessa forma, a comparação mais relevante é o comportamento da resposta em frequência e a região de atenuação, e não apenas o valor absoluto em dB.

## Complexidade computacional

Os três filtros apresentam diferenças importantes em relação à quantidade de operações necessárias para realizar a filtragem.

O **FIR** possui 152 coeficientes. A cada amostra, sua estrutura exige uma quantidade relativamente elevada de operações de multiplicação e acumulação. Como vantagem, sua estrutura não possui realimentação, o que torna o comportamento numérico mais simples e previsível.

O **IIR de 3 SOS** utiliza apenas três seções de segunda ordem, totalizando 15 coeficientes. Dessa forma, sua estrutura é consideravelmente mais compacta que a do FIR para obter uma resposta de filtragem semelhante.

O **notch IIR** é ainda mais compacto, pois utiliza somente uma seção de segunda ordem e cinco coeficientes. Isso permite implementar uma atenuação bastante seletiva com uma quantidade pequena de parâmetros.

Neste projeto não foram obtidas medições numéricas do número de ciclos de CPU utilizando o contador de ciclos do Cortex-M4. Portanto, a comparação de complexidade apresentada é baseada na estrutura dos filtros e na quantidade de coeficientes, e não em uma medição direta de tempo de execução.

## Vantagens, desvantagens e possíveis aplicações

| Implementação | Vantagens | Desvantagens | Possíveis aplicações |
|---|---|---|---|
| **FIR `float32`** | Estrutura simples, sem realimentação e comportamento estável | Grande quantidade de coeficientes e operações | Processamento de áudio, equalização e filtragem geral |
| **FIR Q15** | Representação compacta e adequada para processamento em ponto fixo | Introduz efeitos de quantização e exige atenção à escala dos sinais | Sistemas embarcados com recursos limitados |
| **IIR `float32`** | Poucos coeficientes para obter uma resposta de filtragem semelhante | Possui realimentação e requer maior atenção à estabilidade numérica | Processamento de áudio, instrumentação e sistemas embarcados |
| **Notch IIR** | Estrutura extremamente compacta e alta seletividade | Atua principalmente em uma faixa estreita e requer conhecimento da frequência da interferência | Rejeição de interferências senoidais específicas |

## Conclusão

Os resultados mostraram que as três estruturas implementadas foram capazes de atenuar a interferência localizada em aproximadamente **2465 Hz**. O FIR apresentou uma faixa de rejeição mais ampla e permitiu a comparação entre ponto flutuante e ponto fixo. As duas representações apresentaram resultados semelhantes nos testes realizados.

O IIR apresentou uma estrutura mais compacta, utilizando três seções de segunda ordem para obter uma resposta semelhante à do FIR, enquanto o filtro notch utilizou apenas uma seção de segunda ordem e apresentou uma região de rejeição mais estreita e direcionada à frequência de interesse.

De forma geral, os resultados obtidos no Cortex-M4 foram coerentes com as simulações realizadas em Python, principalmente quanto às regiões de atenuação e ao comportamento dos filtros. As diferenças observadas nos níveis absolutos dos espectros podem ser visualizadas nos resultados experimentais, mas não alteram a conclusão de que as três implementações realizaram a filtragem esperada.
